# Exploratory Data Analysis (EDA)  

## Introduction

This notebook presents an Exploratory Data Analysis (EDA) of registered workers from the eShram dataset.

The analysis focuses on understanding the demographic, educational, occupational, and geographic characteristics of the workers.

> **Note:** The current analysis uses a 300-record development sample for exploration and testing. It does not represent the complete eShram dataset.

**Education Analysis**

```python
Education_counts = df['educationQualification'].value_counts().sort_values(ascending=False)
Education_counts_percentage = (Education_counts*100)/df.shape[0]
print(Education_counts)
print("=============================")
print(Education_counts_percentage)
```

**Education x Gender Analysis**

```python
Education_gender = (
    df.groupby(['educationQualification', 'gender'])
      .size()
      .reset_index(name='count')
)

Education_gender['percentage'] = (
    Education_gender['count']
    / Education_gender.groupby('educationQualification')['count'].transform('sum')
    * 100
).round(2)

Education_gender
```

**Occupation Analysis**

```python
occuption_counts = df['primaryOccupation'].value_counts().head()
occuption_counts
```

**Occupation concentration**

```python
occupation_counts = (
    df['primaryOccupation'].value_counts() * 100 / df.shape[0]
).round(2)

occupation_counts.head()
```

```python
df['primaryOccupation'].value_counts().head().sum()/df.shape[0]*100

# the top 5 occupations reserved 76% of 300-record development sample.
```

**Age distribution**

```python
import pandas as pd
age_bins = [16, 24, 34, 44, 54, 59]
age_labels = ['16-24', '25-34', '35-44', '45-54', '55-59']

df['age_group'] = pd.cut(
    df['age'],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

df['age_group'].value_counts().sort_values(ascending=False)
```

**Age × Gender Analysis**

```python
Age_gender = (
    df.groupby(['age_group', 'gender'])
      .size()
      .reset_index(name='count')
)

Age_gender
```

**Differently Abled Analysis**


```python
diffAbled_counts = df['differentlyAbledStatus'].value_counts()
diffAbled_percentage = diffAbled_counts/df.shape[0]*100

print(diffAbled_counts)
print(diffAbled_percentage)
```

```python
diffAbled_analysis = pd.DataFrame({
    'Status': diffAbled_counts.index,
    'Count': diffAbled_counts.values,
    'Percentage': diffAbled_percentage.values
})

diffAbled_analysis
```

**State-level workforce distribution**

```python
state_workforce_count = df['currentStateName'].value_counts().head()
state_workforce_percentage = (state_workforce_count/df.shape[0]*100).round(2)
state_workforce_percentage
```

```python
State_Workforce_analysis = pd.DataFrame(
    {
        'currentStateName': state_workforce_count.index,
        'count': state_workforce_count.values,
        'Percentage': state_workforce_percentage.values
    }
)

State_Workforce_analysis 
```
